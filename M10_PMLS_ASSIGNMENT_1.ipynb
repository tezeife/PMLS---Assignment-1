```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/skin clinic campaign.csv")
df.head()
```

```python
df["Response"] = df["Response_to_Campaign"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df.head()
```

Gender vs Campaign Response

```python
gender_response = (
    df.groupby("Gender")["Response"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

gender_response.columns = [
    "Gender",
    "Response Rate (%)"
]

gender_response
```

Age Group vs Campaign Response

```python
age_group_response = (
    df.groupby("AgeGroup")["Response"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)
age_group_response.columns = [
    "Age Group",
    "Response Rate (%)"
]
age_group_response
```

Purchase in Last Quarter vs Campaign Response

```python
purchase_response = (
    df.groupby("Purchase_Last_Quarter")["Response"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

purchase_response.columns = [
    "Purchased Last Quarter",
    "Response Rate (%)"
]

purchase_response
```

Product Usage vs Campaign Response

```python
df["Product_Usage"] = pd.cut(
    df["Unique_Products_Purchased"],
    bins=[0, 4, 8, np.inf],
    labels=["1-4", "5-8", ">8"]
)
```

```python
product_response = (
    df.groupby(
        "product_Usage",
        observed=False
    )["Response"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

product_response.columns = [
    "Product Usage",
    "Response Rate (%)"
]

product_response
```
